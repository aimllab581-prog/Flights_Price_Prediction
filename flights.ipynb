```python
import pandas as pd
```

```python
df = pd.read_csv("Clean_Dataset.csv")
```

```python
df.head(10)
```

```python
df = df[["airline"	,"source_city"	,"departure_time"	,"stops"	,"arrival_time","destination_city"	,"class"	,"duration"	,"days_left","price"]]
```

```python
df
```

```python
df = df.drop_duplicates()
```

```python
df
```

```python
df = df.dropna()
```

```python
df
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
le = LabelEncoder()
```

```python
from sklearn.preprocessing import OrdinalEncoder
```

```python
categorical_cols = df.select_dtypes(include=['object', 'category']).columns
```

```python
encoder = OrdinalEncoder()
df[categorical_cols] = encoder.fit_transform(df[categorical_cols]).astype(int)
```

```python
df
```

```python
X = df.iloc[:, :-1]
```

```python
X
```

```python
y = df['price']
```

```python
from sklearn.tree import DecisionTreeRegressor
```

```python
model = DecisionTreeRegressor()
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X,y,test_size=0.2)
```

```python
X_train
```

```python
X_test
```

```python
model.fit(X_train,Y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
r2_score(Y_test,y_pred)
```

```python
mean_squared_error(Y_test,y_pred)
```

```python
mean_absolute_error(Y_test,y_pred)
```

```python
results = pd.DataFrame({'Actual': Y_test, 'Predicted': y_pred})
results['Absolute_Error'] = (results['Actual'] - results['Predicted']).abs()

# Sort to see your 5 biggest misses
print(results.sort_values(by='Absolute_Error', ascending=False).head())
```

```python
results
```

```python
results.plot(x="Actual",y="Predicted",kind="scatter",colormap="winter",c="Actual")
```
